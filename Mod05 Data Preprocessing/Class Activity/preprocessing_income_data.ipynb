```python
import pandas as pd
import numpy as np

df = pd.read_csv("./DirtyData.csv")
df.head()
```

### Step 1: Inspect the data before doing anything

```python
print(df.isna().sum())
print()
print(df["gender"].value_counts())
```

### Step 2: Handle missing valeus by reconstruction

```python
# columns WHERE income is na, tax is NOT na, then grab tax * (100/15), and put that new value into income
df["income"] = np.where(df["income"].isna() & df["tax_15"].notna(), df["tax_15"]*(100/15), df["income"])

# same as above but work backwards to get tax column from income
df["tax_15"] = np.where(df["tax_15"].isna() & df["income"].notna(), df["income"]*(15/100), df["tax_15"])
```

### Step 3: Fix negative incomes

```python
df["income"] = df["income"].abs()
df["tax_15"] = df["tax_15"].abs()

print(f"Negative incomes remaining: {(df["income"] < 0).sum()}")
```

### Step 4: standardize gender with judgment

```python
df["gender"].value_counts()

gender_map = {
    "Men" : "Male", "Man":"Male",
    "Woman":"Female", "Women":"Female"
}

df["gender"] = df["gender"].replace(gender_map)
df.head(20)
```

### Step 5: Normalize Income (min-max)

```python
_min = df["income"].min()
_max = df["income"].max()

income_range = (_max - _min)
df["income_norm"] = (df["income"] - _min) / income_range
```

### Step 6: Final check and save



```python
df.to_csv("./CleanData.csv", index= False)
```
